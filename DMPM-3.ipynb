```python
import matplotlib.pyplot as plt
students = ["A", "B", "C", "D", "E"]
marks = [70, 85, 60, 90, 75]
plt.bar(students, marks)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
students = ["A", "B", "C", "D", "E"]
marks = [70, 85, 60, 90, 75]
sns.barplot(x=students, y=marks)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```
