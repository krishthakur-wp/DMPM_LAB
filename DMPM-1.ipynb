# Experiment 1: Statistical Measures

**Aim:** To calculate mean, median, mode, variance, standard deviation and quartiles using Python.

**Theory:** These measures summarize the centre and spread of numerical data. Mean is the arithmetic average, median is the middle value, mode is the most frequent value, variance measures spread, and standard deviation is the square root of variance.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import importlib.util, subprocess, sys
if importlib.util.find_spec('mlxtend') is None:
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', 'mlxtend'])
```

```python
from statistics import mean, median, multimode, variance, stdev

data = [10, 12, 12, 15, 18, 20, 20, 20, 25, 28]

print("Dataset:", data)
print("Mean:", mean(data))
print("Median:", median(data))
print("Mode:", multimode(data))
print("Sample variance:", round(variance(data), 2))
print("Sample standard deviation:", round(stdev(data), 2))
print("Quartiles (Q1, Q2, Q3):", np.percentile(data, [25, 50, 75]))
print("Minimum:", min(data))
print("Maximum:", max(data))
print("Range:", max(data) - min(data))
```

```python
# The same statistical summary using Pandas
series = pd.Series(data, name="Values")
print(series.describe())
```

**Result:** The central tendency and dispersion of the given dataset were calculated successfully.
